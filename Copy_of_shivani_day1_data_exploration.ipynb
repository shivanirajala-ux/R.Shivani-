```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Agriculture Dataset
data = {
    "Rainfall": [120, 150, 100, 180, 90, 200, 130, 160, 110, 190],
    "Temperature": [28, 30, 32, 27, 35, 26, 29, 31, 34, 25],
    "Soil_Moisture": [40, 55, 35, 70, 30, 80, 45, 65, 38, 75],
    "Crop_Yield": [45, 60, 38, 75, 30, 85, 50, 68, 35, 80]
}

df = pd.DataFrame(data)

# Display dataset
print(df)

# Dataset shape
print("Shape:", df.shape)

# Check missing values
print(df.isnull().sum())

# Box plot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df)

plt.title("Agriculture Data - Box Plot")
plt.show()
```
