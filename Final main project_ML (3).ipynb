```python

## import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (mean_absolute_error, mean_squared_error, r2_score,
                             accuracy_score, confusion_matrix)
data = {
    "Study Hours":    [2, 3, 5, 1, 6, 4, 2, 7, 3, 8, 5, 1],
    "Attendance":     [60, 65, 75, 40, 85, 70, 50, 90, 55, 95, 80, 35],
    "Previous Marks": [45, 50, 60, 30, 70, 55, 35, 80, 40, 85, 65, 25],
    "Final Marks":    [48, 52, 65, 35, 78, 60, 40, 88, 45, 92, 72, 30],
    "Result":         ["Fail", "Pass", "Pass", "Fail", "Pass", "Pass",
                       "Fail", "Pass", "Fail", "Pass", "Pass", "Fail"],
}
df = pd.DataFrame(data)


print("=" * 60)
print("DATASET INSPECTION")
print("=" * 60)
print(df, "\n")
print("Shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
print("\nSummary statistics:\n", df.describe())
print("\nResult counts:\n", df["Result"].value_counts())


features = ["Study Hours", "Attendance", "Previous Marks"]
X = df[features]


new_student = pd.DataFrame([[5, 80, 65]], columns=features)


print("\n" + "=" * 60)
print("TASK 1: LINEAR REGRESSION (Predict Final Marks)")
print("=" * 60)


y_reg = df["Final Marks"]


Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X, y_reg, test_size=0.25, random_state=42)
print("Training samples:", len(Xr_train), "| Testing samples:", len(Xr_test))


reg = LinearRegression()
reg.fit(Xr_train, yr_train)


yr_pred = reg.predict(Xr_test)
mae = mean_absolute_error(yr_test, yr_pred)
rmse = np.sqrt(mean_squared_error(yr_test, yr_pred))
r2 = r2_score(yr_test, yr_pred)
print("Actual marks   :", yr_test.values)
print("Predicted marks:", np.round(yr_pred, 2))
print(f"MAE = {mae:.2f} | RMSE = {rmse:.2f} | R2 = {r2:.3f}")


pred_marks = reg.predict(new_student)[0]
print("\nNew student: Study Hours=5, Attendance=80, Previous Marks=65")
print(f"Predicted Final Marks: {pred_marks:.2f}")


print("\n" + "=" * 60)
print("TASK 2: LOGISTIC REGRESSION (Predict Pass / Fail)")
print("=" * 60)


label_map = {"Fail": 0, "Pass": 1}
label_names = {0: "FAIL", 1: "PASS"}          # replaces if-else
y_clf = df["Result"].map(label_map)


Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X, y_clf, test_size=0.25, random_state=42, stratify=y_clf)
print("Training samples:", len(Xc_train), "| Testing samples:", len(Xc_test))


clf = LogisticRegression(max_iter=1000)
clf.fit(Xc_train, yc_train)


yc_pred = clf.predict(Xc_test)
acc = accuracy_score(yc_test, yc_pred)
print("Actual    :", yc_test.values)
print("Predicted :", yc_pred)
print("Accuracy  :", acc)
print("Confusion Matrix:\n", confusion_matrix(yc_test, yc_pred, labels=[0, 1]))


pred_label = clf.predict(new_student)[0]
pred_prob = clf.predict_proba(new_student)[0]
print("\nNew student: Study Hours=5, Attendance=80, Previous Marks=65")
print(f"Predicted Result: {label_names[pred_label]}")
print(f"Probability -> Fail: {pred_prob[0]:.2f}, Pass: {pred_prob[1]:.2f}")


reg_full = LinearRegression().fit(X, y_reg)
clf_full = LogisticRegression(max_iter=1000).fit(X, y_clf)


def ask(prompt, low, high):
    """Keep asking until a number is typed, then clip it into the valid range."""
    while True:
        try:
            return float(np.clip(float(input(prompt)), low, high))
        except ValueError:
            print("  Invalid input. Please enter a number.")


print("=" * 60)
print(" LIVE PREDICTION")
print("=" * 60)

again = "y"
while again == "y":
    hours = ask("\nStudy Hours per day (0-12)  : ", 0, 12)
    attend = ask("Attendance % (0-100)         : ", 0, 100)
    prev = ask("Previous Marks (0-100)        : ", 0, 100)

    student = pd.DataFrame([[hours, attend, prev]], columns=features)
    marks = np.clip(reg_full.predict(student)[0], 0, 100)
    label = clf_full.predict(student)[0]
    prob = clf_full.predict_proba(student)[0]

    print("\n" + "-" * 50)
    print(f" Predicted Final Marks : {marks:.2f}")
    print(f" Predicted Result      : {label_names[label]}")
    print(f" Probability           : Pass {prob[1]*100:.1f}% | Fail {prob[0]*100:.1f}%")
    print("-" * 50)

    again = input("\nPredict for another student? (y/n): ").strip().lower()

print("Thank you!")
```
